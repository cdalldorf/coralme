```python
# use the coralme_old2 environment for this
import coralme
import pandas
from coralme.solver.solver import ME_NLP
import cobra
import numpy
from coralme.core.optimization import _check_options
from coralme.core.optimization import construct_lp_problem

#from coralme.core.optimization import fva
import matplotlib.pyplot as plt
plt.rcParams['svg.fonttype'] = 'none'
pandas.set_option('display.max_colwidth', None)
import inspect

import pandas as pd
import seaborn as sns
import numpy as np
from sklearn.metrics import r2_score
import math
import os
import pickle
import pandas as pd

coralme_dir = '/home/chris/zuniga/coralme/' #'../'

# build pathway reference using Carlos mapping
models_carlos_df = pd.read_csv(os.path.join(coralme_dir, 'data', 'external', 'geneIDsMapping_PA14_vs_Strains_SubsystemInfo.csv'), index_col = 0)
```

```python
# check for solutions, run if nonexistant
base_dir = os.path.join(coralme_dir, 'species_files', 'Pseudomonas_files')
for f in os.listdir(os.path.join(base_dir, 'individual_species')):
    if 'Reference' in f: continue

    # load and run model
    print(f+' working...', end = '')
    out_path = os.path.join(base_dir, 'individual_species', f, 'outputs')
    M_model_path = os.path.join(out_path, 'pathway_annotated_M_model.sbml')
    M_model = cobra.io.sbml.read_sbml_model(M_model_path)
    M_sol_path = os.path.join(base_dir, 'individual_species', f, 'outputs', 'M_flux_solution.pkl')
    M_flux_path = os.path.join(base_dir, 'individual_species', f, 'outputs', 'M_flux_ref_solution_dict.pkl')
    M_flux_path2 = os.path.join(base_dir, 'individual_species', f, 'outputs', 'M_flux_carlos_solution_dict.pkl')
    
    model_ = coralme.core.model.MEModel.from_cobra(M_model)
    model_.optimize()
    solution = model_.solution
    
    # if there is a solution, create pathway to flux and annotation, save them
    if solution:
        print('solution found!')
        rxns = []
        fluxes = []
        pathways = []
        rxns2 = []
        fluxes2 = []
        pathways2 = []
        for rxn, flux in model_.solution.fluxes.items():
            M_rxn = M_model.reactions.get_by_id(rxn)
            if 'ref_pathways' in M_rxn.annotation:
                rxns.append(rxn)
                fluxes.append(flux)
                pathways.append(M_rxn.annotation['ref_pathways'])
            paths_temp = set()
            for gene in M_rxn.gpr.genes:
                check = models_carlos_df[models_carlos_df[f] == gene]
                if len(check) > 0:
                    sub = models_carlos_df[models_carlos_df[f] == gene].Subsystem.values[0]
                    paths_temp.add(sub)
            if len(paths_temp) > 0:
                paths_temp = list(paths_temp)
                paths_temp.sort()
                rxns2.append(rxn)
                fluxes2.append(flux)
                pathways2.append(paths_temp)
        flux_df = pd.DataFrame(index = rxns)
        flux_df['flux'] = fluxes
        flux_df['pathways'] = pathways
        flux_df.to_pickle(M_flux_path)
        flux_df2 = pd.DataFrame(index = rxns2)
        flux_df2['flux'] = fluxes2
        flux_df2['pathways'] = pathways2
        flux_df2.to_pickle(M_flux_path2)
        
        # save off solution
        pickle_out = open(M_sol_path, 'wb')
        pickle.dump(solution, pickle_out)
        pickle_out.close()
    else:
        print('no solution :(')
        # save something bogus to prevent rerunning this
        pickle_out = open(M_flux_path, 'wb')
        pickle.dump(False, pickle_out)
        pickle_out.close()
        
        # save off solution
        pickle_out = open(M_sol_path, 'wb')
        pickle.dump(False, pickle_out)
        pickle_out.close()
```
